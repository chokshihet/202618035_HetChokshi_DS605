#### Task 2- Vectorized Arithmatic and Linear Algebra

```python
M1 = np.array([[4,7,2],
               [3,6,1],
               [2,5,8]])

M2 = np.array([[1,2,3],
               [0,4,5],
               [1,0,6]])

addition = M1 + M2
elem_mult = M1 * M2
matmul = M1 @ M2

print("--- Matrix Operations ---")
print(f"Addition (M1 + M2):\n{addition}\n")
print(f"Element-wise Multiplication (M1 * M2):\n{elem_mult}\n")
print(f"Matrix Multiplication (M1 @ M2):\n{matmul}\n")

M1_transpose = M1.T
det_M1 = np.linalg.det(M1)

print("Matrix Properties: ")
print(f"Transpose of M1:\n{M1_transpose}")
print(f"Determinant of M1: {det_M1:.4f}\n")

if not np.isclose(det_M1, 0):
    M1_inv = np.linalg.inv(M1)
    identity_calc = M1 @ M1_inv
    
    # Verify M1 @ M1_inv is approximately Identity Matrix I
    is_identity = np.allclose(identity_calc, np.eye(3))
    
    print("--- Matrix Inverse & Verification ---")
    print(f"M1 Inverse:\n{M1_inv}\n")
    print(f"M1 @ M1^-1:\n{np.round(identity_calc, 4)}")
    print(f"Is M1 @ M1^-1 equal to Identity Matrix? {is_identity}")
else:
    print("Matrix is singular and non-invertible.")
```

#### Task 3: Normal Distribution and Histogram


```python
import matplotlib.pyplot as plt

chosen_mean = 50
chosen_std = 10
num_samples = 1000

data = np.random.normal(loc=chosen_mean, scale=chosen_std, size=num_samples)

sample_mean = np.mean(data)
sample_std = np.std(data)

print("Distribution Parameters: ")
print(f"Chosen mean: {chosen_mean} | sample mean: {sample_mean:.4f}")
print(f"Chosen std: {chosen_std} | Sample std: {sample_std:.4f}")

plt.figure(figsize=(8,5))
plt.hist(data, bins=35, edgecolor='blue', color='steelblue', alpha=0.7, density=True)

plt.axvline(chosen_mean, color='red', linestyle='--', linewidth=2, label=f'Chosen Mean ({chosen_mean})')
plt.axvline(sample_mean, color='orange', linestyle=':', linewidth=2, label=f'Sample Mean ({sample_mean:.2f})')

plt.title("Normal Distribution Sample Histogram", fontsize=14)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Probability Density", fontsize=12)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
